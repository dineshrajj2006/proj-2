```python
movies=pd.read_csv("movies_cleaned.csv")
cast=pd.read_csv("cast_cleaned.csv")
crew=pd.read_csv("crew_cleaned.csv")
genres=pd.read_csv("genres_cleaned.csv")
movie_genres=pd.read_csv("movie_genres_cleaned.csv")
movie_keywords=pd.read_csv("movie_keywords_cleaned.csv")
```

```python
query = """
SELECT
    m.title,
    m.release_date
FROM movies AS m
JOIN movie_genres AS mg
    ON m.movie_id = mg.movie_id
JOIN genres AS g
    ON mg.genre_id = g.genre_id
WHERE g.genre_name = 'Action'
  AND m.release_date >= '2016-01-01'
ORDER BY m.release_date ;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
query = """
SELECT
    m.title,
    m.revenue,
    g.genre_name
FROM movies AS m
JOIN movie_genres AS mg
    ON m.movie_id = mg.movie_id
JOIN genres AS g
    ON mg.genre_id = g.genre_id
WHERE m.revenue IS NOT NULL
ORDER BY m.revenue DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```

```python

query = """
SELECT
    g.genre_name,
    AVG(m.budget) AS average_budget,
    AVG(m.revenue) AS average_revenue
FROM movies AS m
JOIN movie_genres AS mg
    ON m.movie_id = mg.movie_id
JOIN genres AS g
    ON mg.genre_id = g.genre_id
GROUP BY g.genre_id, g.genre_name
ORDER BY average_revenue DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python

query = """
SELECT
    actor_name,
    COUNT(DISTINCT movie_id) AS movie_count
FROM cast
GROUP BY person_id, actor_name
ORDER BY movie_count DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```

```python

query = """
SELECT
    person_id,
    person_name,
    COUNT(DISTINCT movie_id) AS movie_count
FROM crew
WHERE job = 'Director'
GROUP BY person_id, person_name
HAVING COUNT(DISTINCT movie_id) > 3
ORDER BY movie_count DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python

query = """
SELECT
    keyword_name,
    COUNT(DISTINCT movie_id) AS movie_count
FROM movie_keywords
GROUP BY keyword_id, keyword_name
ORDER BY movie_count DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```

```python

query = """
SELECT
    title,
    budget,
    revenue
FROM movies
WHERE budget > (
    SELECT AVG(budget)
    FROM movies
    WHERE budget > 0
)
AND revenue < (
    SELECT AVG(revenue)
    FROM movies
    WHERE revenue > 0
)
AND budget > 0
AND revenue > 0
ORDER BY budget DESC;
"""

result = pd.read_sql_query(query, conn)
result

```

```python

query = """
SELECT DISTINCT
    c.actor_name,
    m.title AS movie_title,
    cr.person_name AS director_name
FROM `cast` AS c
JOIN crew AS cr
    ON c.movie_id = cr.movie_id
JOIN movies AS m
    ON c.movie_id = m.movie_id
WHERE cr.job = 'Director'
  AND cr.person_name = 'Christopher Nolan'
ORDER BY m.title, c.actor_name;
"""

result = pd.read_sql_query(query, conn)
result
```

```python

query = """
SELECT
    g.genre_name,
    AVG(m.vote_average) AS average_rating,
    COUNT(DISTINCT m.movie_id) AS movie_count
FROM movies AS m
JOIN movie_genres AS mg
    ON m.movie_id = mg.movie_id
JOIN genres AS g
    ON mg.genre_id = g.genre_id
GROUP BY g.genre_id, g.genre_name
HAVING COUNT(DISTINCT m.movie_id) >= 20
ORDER BY average_rating DESC
LIMIT 1;
"""

result = pd.read_sql_query(query, conn)
result
```

```python

query = """
SELECT
    m.title,
    COUNT(DISTINCT mk.keyword_id) AS keyword_count
FROM movies AS m
JOIN movie_keywords AS mk
    ON m.movie_id = mk.movie_id
GROUP BY m.movie_id, m.title
ORDER BY keyword_count DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```
